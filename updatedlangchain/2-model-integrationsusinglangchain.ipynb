```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
model=ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite"
)
model.invoke("what is the model name and model provider name you are using")
```

Streaming the output while generating


```python
model.invoke("why do parrots have different colors").content
#it is genterating output at final all the information together
```

```python
#model.stream("why do parrots have different colors")

for chunk in model.stream("why do parrots have different colors"):
    print(chunk.content)
#it is printing one by one while generating
```

Batch- collection of requests -> benefits improves performance and reduces costs

```python
responses=model.batch(
    ["who are you",
    "what is meant by food diet",
    "why do parrots talk?"],

    config={
        'max_concurrency':5 #limits the calls to model(parallel calls)
    }
    )
for response in responses:
    print(response.content)
    print("----------")
```
