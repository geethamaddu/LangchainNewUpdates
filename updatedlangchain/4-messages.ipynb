Messages
-It is type of datastructure that provides context to LLM. They respresent the state of converstaion when interacting with LLMs. These are models input and ouput.

it have:

    role- identifies the type message(system, user, ai, tool)

    content- actual message
    
    metadata- optional fields like response infor, message id, token usage



```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

from langchain.chat_models import init_chat_model
model=init_chat_model("groq:llama-3.1-8b-instant")
res=model.invoke("what is the model name and model provider name you are using")
res
```

Message Prompt:

you can pass list of messages which are list of message objects

Messages Types: System message, Ai message, Human message, Tool message

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage
#system message-> instruction how llm to behave
#human message-> userinput that can have text,image audio, file
#ai message-> llm output , tool call information
#tool message-> tool calls output 

message=[
    SystemMessage(content="You are a poetry expert."),
    HumanMessage(content="write a poem on artificial intelligence.")
]
model.invoke(message)
```

```python
from langchain.messages import SystemMessage, HumanMessage

sys_text= """
You are a poetry expert.
it should minimum 6 to 7 lines.
Always provide great poetry.
"""
sys_msg=SystemMessage(content=sys_text)

message=[
    sys_msg,
    HumanMessage(content="write a poem on restapis.")
]
model.invoke(message)
```

```python
human_msg=HumanMessage(
    content="Write poem on ai",
    name="phani",
    id="msg_123"
)

message=[
    sys_msg,
    human_msg
]
model.invoke(message)
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage
ai_message=AIMessage(
    content="",
    tool_calls=[{
        "name":"get_weather",
        "args":{"location":"San Francisco"},
        "id":"call_123"
    }]
)

weather_result="Sunny, 72F"

tool_message=ToolMessage(
    content=weather_result,
    tool_call_id="call_123"
)

messages=[
    HumanMessage("What is the weather in San Francisco"),
    ai_message,
    tool_message
]
response=model.invoke(messages)
response
```
